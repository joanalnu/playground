# Calculando $\pi$

```python
import random
```

```python
num_dardos = 1_000_000_000
```

```python
print(f"Vamos a lanzar {num_dardos} dardos para calcular π")
```

```python
dardos_dentro = 0
```

```python
for i in range(num_dardos):

    # generar coordenadas aleatorias entre -1 y 10
    x = random.uniform(-1, 1)
    y = random.uniform(-1, 1)

    # Calculamos la distancia desde el centro (0, 0) hasta el punto (x, y)
    # Usamos el teorema de Pitágoras: distancia = √(x² + y²)
    distancia = (x**2 + y**2) **0.5

    if distancia <= 1:
        dardos_dentro = dardos_dentro +1
```

```python
pi_aproximado = 4 * (dardos_dentro / num_dardos)
```

```python
print(pi_aproximado)
```
